## 1. Introduction to Dimensionality Reduction

Dimensionality reduction is a technique used to reduce the number of features or dimensions in a dataset while retaining as much relevant information as possible. It is an important part of machine learning because datasets with many features can increase computational requirements and make analysis more difficult.

This chapter explores dimensionality reduction techniques, including Principal Component Analysis (PCA), Linear Discriminant Analysis (LDA), and t-Distributed Stochastic Neighbor Embedding (t-SNE). These methods have different purposes, ranging from reducing the number of features to improving class separation and visualizing high-dimensional data.

Dimensionality reduction is useful for several reasons:

- **Simplifying data:** Reducing the number of features makes high-dimensional datasets easier to analyze and visualize.
- **Reducing computational costs:** Fewer features can reduce the computational resources and time needed for model training.
- **Improving model performance:** High-dimensional data can increase the risk of overfitting when a model learns noise instead of meaningful patterns.
- **Retaining relevant information:** The goal is to produce a simpler representation of the data while preserving important information.

## 4. Theoretical Foundations of Dimensionality Reduction

Dimensionality reduction transforms high-dimensional data into a representation with fewer dimensions. The goal is to simplify the dataset while retaining useful information.

The chapter discusses PCA, LDA, and t-SNE. PCA focuses on capturing variance in the dataset, LDA focuses on separating known classes, and t-SNE is used to visualize high-dimensional data in a lower-dimensional space.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_wine
import warnings

np.random.seed(2024)

warnings.simplefilter(
    action="ignore",
    category=FutureWarning
)

wine = load_wine()

df_wine = pd.DataFrame(
    data=wine.data,
    columns=wine.feature_names
)

target_wine = wine.target

display(df_wine.head(10))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
import matplotlib.pyplot as plt

pca_pipeline = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "pca",
        PCA(n_components=2)
    )
])

X_pca = pca_pipeline.fit_transform(df_wine)
```

```python
from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis
)

X_wine, y_wine = wine.data, wine.target

lda_pipeline_wine = Pipeline([
    ("scaler", StandardScaler()),
    (
        "lda",
        LinearDiscriminantAnalysis(n_components=2)
    )
])

X_lda_wine = lda_pipeline_wine.fit_transform(
    X_wine,
    y_wine
)
```

```python
plt.figure(figsize=(10, 8))

shapes = ["o", "^", "D"]
colors = ["r", "g", "b"]

for i, (shape, color) in enumerate(
    zip(shapes, colors)
):
    mask = y_wine == i

    plt.scatter(
        X_lda_wine[mask, 0],
        X_lda_wine[mask, 1],
        c=color,
        marker=shape,
        edgecolor="black",
        label=wine.target_names[i]
    )

plt.xlabel("First LDA Component")
plt.ylabel("Second LDA Component")
plt.title("Wine Dataset - LDA Components")
plt.legend()
plt.show()
```

```python
from sklearn.datasets import load_digits

digits = load_digits()
```

from sklearn.manifold import TSNE

tsne_pipeline = Pipeline([

        "tsne",

```python
plt.figure(figsize=(10, 8))

scatter = plt.scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    c=digits.target,
    cmap="Paired",
    label=digits.target
)

plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.title("Digits Dataset - t-SNE Visualization")
```

```python
legend_elements = [
    plt.Line2D(
        [0], [0],
        marker="o",
        color="w",
        markerfacecolor=plt.cm.Paired(i / 9),
        label=str(i),
        markersize=10
    )
    for i in range(10)
]

plt.legend(
    handles=legend_elements,
    title="Digit",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)

plt.show()
```

## 8. Selecting the Right Technique

The choice of dimensionality reduction technique depends on the objective of the analysis.

- **PCA:** Suitable for reducing features while capturing as much variance as possible.
- **LDA:** Suitable for supervised classification tasks where class separation is important.
- **t-SNE:** Suitable for visualizing complex, high-dimensional data in two or three dimensions.

The characteristics of the dataset and the intended use of the reduced features should be considered before selecting a technique.

## 9. Impact on Model Performance

Dimensionality reduction can reduce computational costs and help address problems associated with high-dimensional data. However, reducing the number of dimensions may also remove useful information.

The effects should therefore be evaluated by comparing models trained on the original features and the reduced features. Performance metrics and the intended purpose of the model should guide the final decision.
