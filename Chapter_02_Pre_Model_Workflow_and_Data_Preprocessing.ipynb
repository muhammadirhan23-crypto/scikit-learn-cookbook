## The Impact of Raw Data on Model Performance

Machine learning algorithms are designed to learn patterns from data. However, when input data contains missing values, outliers, or irrelevant features, the model's ability to generalize to unseen data can decrease.

A model trained on noisy or biased data may produce inaccurate predictions and lead to poor decision-making in real-world applications. Therefore, data preprocessing is an important stage in machine learning because the quality of the input data affects the performance of the model.

This chapter covers techniques for handling missing data, scaling numerical features, encoding categorical variables, creating pipelines, and performing feature engineering.

## Common Data Issues

Some of the most common data quality issues in machine learning include:

- **Missing Data:** Incomplete data may occur because of errors during data collection, system failures, or data corruption. scikit-learn provides methods such as SimpleImputer() and KNNImputer() to handle missing values.
- **Outliers:** Outliers can distort statistical analyses and produce misleading results. Z-score analysis and RobustScaler() can help address their impact.
- **Categorical Variables:** Many machine learning algorithms require numerical input. Categorical data can be transformed using tools such as OneHotEncoder() and LabelEncoder().
- **Feature Scaling:** Features may have different value ranges, so scaling can help prevent some features from dominating the learning process.
- **Data Leakage:** Information that should not be available during training can affect evaluation and make a model appear more effective than it actually is.

## Cleaning and Preparing Data

Data preprocessing involves identifying and addressing issues that may affect machine learning performance. These issues include missing values, outliers, categorical variables, and differences in feature scales.

scikit-learn provides tools such as Pipeline(), ColumnTransformer(), SelectFromModel(), and Recursive Feature Elimination (RFE) to support data preparation and feature selection.

```python
import numpy as np
import pandas as pd

np.random.seed(2024)  # For reproducibility
n_samples = 20
n_features = 10

data = {
    f"Feature{i+1}": np.random.uniform(0, 100, n_samples)
    for i in range(n_features)
}

df = pd.DataFrame(data)

for column in df.columns:
    mask = np.random.random(n_samples) < 0.2
    df.loc[mask, column] = np.nan

display(df)
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="mean")

imputed_data = imputer.fit_transform(df)

imputed_df = pd.DataFrame(
    imputed_data,
    columns=df.columns
)

imputed_df
```

```python
from sklearn.impute import KNNImputer

knn_imputer = KNNImputer(n_neighbors=2)

knn_imputed_data = knn_imputer.fit_transform(df)

knn_imputed_df = pd.DataFrame(
    knn_imputed_data,
    columns=df.columns
)

knn_imputed_df
```

```python
from sklearn.experimental import (
    enable_iterative_imputer
)
from sklearn.impute import IterativeImputer

iterative_imputer = IterativeImputer()

iterative_imputed_data = (
    iterative_imputer.fit_transform(df)
)

iterative_imputed_df = pd.DataFrame(
    iterative_imputed_data,
    columns=df.columns
)

iterative_imputed_df
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(
    iterative_imputed_df
)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=iterative_imputed_df.columns
)

scaled_df
```

```python
from sklearn.preprocessing import MinMaxScaler

minmax_scaler = MinMaxScaler()

minmax_scaled_data = minmax_scaler.fit_transform(
    iterative_imputed_df
)

minmax_scaled_df = pd.DataFrame(
    minmax_scaled_data,
    columns=iterative_imputed_df.columns
)

minmax_scaled_df
```

```python
from sklearn.preprocessing import Normalizer

normalizer = Normalizer()

normalized_data = normalizer.fit_transform(
    iterative_imputed_df
)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=iterative_imputed_df.columns
)

normalized_df
```

## Encoding Categorical Variables

Categorical variables represent discrete values such as categories, labels, or groups. They are divided into nominal variables, which have no inherent ordering, and ordinal variables, which have a clear ordering.

OneHotEncoder() converts nominal categories into binary features. LabelEncoder() assigns an integer to each category, but the assigned integers may introduce unintended ordinal relationships if the categories have no natural order.

The book also discusses ColumnTransformer() as a tool for applying different preprocessing techniques to different columns.
